```python
import pandas as pd

def load_data(file_path):
    try:
        data = pd.read_csv(file_path)
        return data
    except FileNotFoundError:
        print(f"Error: The file at {file_path} was not found.")
        return None 

file_path = '../data/technicals/AAPL_data.csv'
df = load_data(file_path)

print(len(df))
```

```python
size = len(df)

train = df[:int(size*0.7)]
gap = df[int(size*0.7):int(size*0.7125)]
test = df[int(size*0.7125):]
```
